```python
import numpy as np
```

Here are some basic operations on matrices in Numpy. First, generate a randome matrix

```python
matX = np.arange(60).reshape(6,10)
print (matX)
```

```python
# Print the diagonal, upper triangle and lower triangle of matrix X
print(np.diag(matX))
print(np.triu(matX))
print(np.tril(matX))
```

We can also compute the matrix products of two matrices

```python
mat1 = np.array([[2, 3, 4],[1, 2, 4]])
print(mat1)
mat2 = np.array([[0, -1],[3, -4],[1, 2]])
print(mat2)
```

```python
print(mat1 @ mat2)
```

Here is the transpose

```python
mat1_T = mat1.T
print(mat1_T)
```

And the determinant...

```python
matA = np.array([[-1, 3],[1, -1]])

print(np.linalg.det(matA))
```

And the inverse of a matrix, just like the example in class.

```python
matA_inv = np.linalg.inv(matA)
print(matA_inv)
```

Housing Data Example

```python
# Enter the Housing Data matrices

A = np.array([[1, 1500, 3],
              [1, 2000, 4],
              [1, 1800, 3]])
B = np.array([300000, 400000, 330000]).T
x = np.linalg.inv(A)@B
print(x)
```

```python
# And again with the expanded Housing Data matrices

A1 = np.array([[1, 1500, 3],
               [1, 2000, 4],
               [1, 1800, 3],
               [1, 2200, 4],
               [1, 1600, 3],
               [1, 2400, 5],
               [1, 1700, 3]])
B1 = np.array([[300000],
               [400000],
               [330000],
               [440000],
               [310000],
               [480000],
               [320000]])
D = np.linalg.pinv(A1) @ B1
print(D)
```

```python
estimated_prices = A1 @ D
print(estimated_prices)
```

Note that the estimated prices do not exactly match the input prices we have in matrix B1! This is because we've essentially solved for a regression line through the points!
