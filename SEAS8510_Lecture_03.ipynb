Eigendecomposition

Start with an example 2x2 matrix and perform an eigendecomposition. 

```python
import numpy as np
np.set_printoptions(precision=2, suppress=True)
```

```python
A = np.array([[4,1],[2,3]])
print(A)
```

```python
# Print out the results of numpy's eigenvalue/eigenvector calcuation directly to see its format

print(np.linalg.eig(A))
```

```python
# Two arrays are apparent. The first array contains the eigenvalues. The second array contains the eigenvector matrix. 
# REMEMBER: THE EIGENVECTORS ARE THE COLUMNS OF THE EIGENVECTOR MATRIX!!

evals = np.linalg.eig(A)[0]
evecs = np.linalg.eig(A)[1]

print(evals)
print(evecs)
```

Also remember, there are infinitely many scaled solutions of the characteristic equation for each eigenvector. Numpy returns unit-normalized eigenvectors, that is, a nominal eigenvector divided by its L2 norm (Euclidean). 

```python
# Verify that the eigendecomposition returns the original matrix

L = np.diag(evals)
V = evecs

A_new = V @ L @ np.linalg.inv(V)
print(A_new)
```

Here we implement a basic singular value decomposition on a sample matrix.

```python
B = np.array([[1, 2],
              [3, 4],
              [5, 6]])
print("Original matrix B: \n", B)
```

```python
U, S, V_T = np.linalg.svd(B, full_matrices=True)
print("Matrix U (Left singular vectors): \n", U)
print("Diagonal Matrix Sigma (Singular values): \n", S)
print("Matrix V-transpose (Transposed right singular vectors): \n", V_T)
```

NOTE: Using SVD in Numpy has a few features to note:
1) The singular values, S, are returned as a vector not a diagonal matrix. You can make a diagonal matrix using the code below.
2) The right-singular vectors are returned in the transposed matrix, V-transpose. To get actual V, you must transpose again.

```python
Sigma = np.zeros(np.shape(B))
np.fill_diagonal(Sigma, S)
print("Diagonal Matrix Sigma (Singular values): \n", Sigma)

V = V_T.T
print("Matrix V (Right singular vectors): \n", V)
```

```python
# Verify that we can return the original matrix B using the SVD components

B_reconstructed = U @ Sigma @ V_T
print("Matrix B reconstructed: \n", B_reconstructed)
```

```python
# Remember the eigenvectors of AA_T are the left-singular vectors and those of A_TA are the right-singular vectors.
U_eig = np.linalg.eig(B @ B.T)[1]
print(U_eig)

# Remember the rows of V are the singular vectors, but the columns of the eigendecomposition are the vectors.
# Each column of the eigenvectors corresponds to each row of the right-singular vectors.
V_eig = np.linalg.eig(B.T @ B)[1]
print(V_eig)
```

```python
# Finally, the square root of the eigenvalues equals the singular values.

S_eig = np.sqrt(np.linalg.eig(B.T @ B)[0])
print(S_eig)
```
