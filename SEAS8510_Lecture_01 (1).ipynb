```python
# Dot Products and Norms
```

```python
vec = np.array([1,2,3,4,5,6])
vec_mag = np.linalg.norm(vec)
print(vec_mag)
```

```python
vec1 = np.array([5,0])
vec2 = np.array([2,4])
dot_prod = np.dot(vec1,vec2)
print(dot_prod)
```

```python
angle = np.arccos(dot_prod/(np.linalg.norm(vec1)*np.linalg.norm(vec2)))
print(np.degrees(angle))
```

```python
math = np.array([87, 64, 94, 45, 76])
science = np.array([99, 84, 82, 58, 79])
```

```python
x = math - np.mean(math)
y = science - np.mean(science)
```

```python
correlation = np.dot(x,y)/(np.sqrt(np.dot(x, x))*np.sqrt(np.dot(y,y)))
print(correlation)
```

```python
print(np.corrcoef(math, science))
```
